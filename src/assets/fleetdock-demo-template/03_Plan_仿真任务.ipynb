## 为什么使用 Plan

Plan 把多个步骤组成一个任务。SDK 负责顺序执行、完成确认、超时和标准结果，业务代码只描述“要做什么”。


```python
from fleetdock_runtime import FleetDock
from fleetdock import RelativePoint3D
import time

FleetDock.init()
drone = FleetDock.devices.drone_yj_001

connect_response = drone.connect()
print("连接：", connect_response)

unlock_response = drone.unlock()
print("解锁：", unlock_response)

time.sleep(2)
print("状态：", drone.state.snapshot())

```

```python
result = (
    FleetDock.core.plan(drone)
    .wait_until_ready(timeout_s=60.0)
    .unlock()
    .takeoff(altitude=0.8)
    .move(RelativePoint3D(x_m=8.0, y_m=0.0, z_m=10.0))
    .return_to_launch()
    .run(step_timeout_s=120.0)
)

print(result)
print("任务成功：", result.ok)
print("步骤结果：", result.steps)
```

```python
FleetDock.close()
```

## 重点

链式调用只是在声明任务，真正执行发生在 `run()`。应用不需要自己轮询高度、维护步骤状态或拼接厂商命令。
