> 运行方式：从上到下执行代码单元。控制类 Notebook **没有布尔安全开关**；执行控制单元就会把任务提交给 SDK。
> 需要仿真的课程请先在 FleetDock IDEA 中启动仿真，并确认 `drone1`、`drone2` 等设备在线。


![FleetDock IDEA 运行时](assets/runtime.svg)


## 1. 初始化

`fleetdock_runtime.py` 由 FleetDock IDEA 根据当前应用配置生成。它把已经绑定的设备变成
`FleetDock.devices.<别名>`，因此业务代码不需要再次填写端口、连接串或产品型号。


```python
from fleetdock_runtime import FleetDock

FleetDock.init()
FleetDock.info()
```

## 2. 取得标准设备对象

下面直接取得应用中名为 `drone1` 的设备。返回对象是 FleetDock SDK 的 `Drone`，后续命令、事件、
状态、传感器和 Plan 都从这个对象开始。


```python
drone = FleetDock.devices.drone1
print(drone.metadata)
print(drone.profile)
print(drone.diagnostics())
```

## 3. 读取应用配置

`envs` 用于普通环境变量，`routes.from_env()` 可以把 FleetDock IDEA 地图编辑器保存的航点直接转换成 SDK `Route`。


```python
print("APP_NAME =", FleetDock.envs.get("APP_NAME", "未配置"))
print("全部绑定设备 =", FleetDock.info()["devices"])
```

```python
FleetDock.close()
```

## 记住

日常应用代码的最短路径就是：`FleetDock.init()` → `FleetDock.devices` → 调用 SDK → `FleetDock.close()`。
