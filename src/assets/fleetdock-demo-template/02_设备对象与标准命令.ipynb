## 标准命令的价值

无论底层是 PX4、厂商平台还是 Mock，应用面对的都是相同的 `connect / unlock / takeoff / move / return_to_launch`。
SDK 负责把标准命令交给对应适配器。


```python
from fleetdock import DeviceFactory, DeviceMetadata, RelativeMove3D

drone = (
    DeviceFactory.create()
    .mock_drone()
    .metadata(DeviceMetadata(device_id="teaching-drone", name="教学无人机"))
    .build()
)

print(drone.connect())
print(drone.unlock())
print(drone.takeoff(altitude=3.0))
print(drone.move(RelativeMove3D(forward_m=2.0, right_m=1.0, up_m=0.0)))
print(drone.return_to_launch())
drone.close()
```

## 关键 API

- `DeviceFactory`：用链式构造器创建标准设备。
- `RelativeMove3D`：明确表达“向前、向右、向上”的米制位移。
- `StandardResponse`：所有标准命令都统一返回 `ok / code / message / data`。

即时命令的成功表示命令被适配器接受；需要确认完整动作时，下一课使用 `Plan`。
